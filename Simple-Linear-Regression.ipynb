```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# loading data set
dataset = pd.read_csv(r'./Practice-data-sets/Salary_Data.csv')
```

```python
dataset
```

```python
dataset.shape
```

```python
# Feature selection
```

```python
X = dataset.iloc[:,:-1].values
```

```python
X
```

```python
Y =  dataset.iloc[:,-1].values
```

```python
Y
```

```python
type(Y)
```

```python
# Split datset set into train and test data
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=0) # 80% training, 20% test
```

```python
x_test, x_train
```

```python
x_test.shape
```

```python
# Fit Linear Regression

from sklearn.linear_model import LinearRegression
```

```python
# Linear Regression is an algorithm and Regressor is model
```

```python
regressor = LinearRegression()
```

```python
regressor.fit(x_train, y_train) # training the model by providing the input values and thier corresponding output values
```

```python
# Predicting the results for the test data set
y_predicted = regressor.predict(x_test)
```

```python
# visualizing the trainind set results

plt.scatter(x_train, y_train, color='red') # actual data
```

```python
plt.plot(x_train, regressor.predict(x_train), color="blue") # predict regression line
```

```python
plt.title("Salary vs Experience")
```

```python
plt.scatter(x_train, y_train, color='red') # actual
plt.scatter(x_train, regressor.predict(x_train), color='purple') # predicted
plt.plot(x_train, regressor.predict(x_train), color='green') # showing best fit line.
plt.xlabel("Years of Exp")
plt.ylabel("Salary")
plt.show()
```

```python
plt.scatter(x_train, y_train, color='red', label="Actual data") # actual
plt.scatter(x_train, regressor.predict(x_train), color='purple', label="Predicted data") # predicted
plt.plot(x_train, regressor.predict(x_train), color='green', label="Best Fit line") # showing best fit line.
plt.xlabel("Years of Exp")
plt.ylabel("Salary")
plt.legend()
plt.show()
```

```python
# output coefficients of Linear model
```

```python
print(f"Coefficeint (m): {regressor.coef_}")
```

```python
print(f"Intercept (c): {regressor.intercept_}")
```

```python
print(f"Linear regression equation is : y = {regressor.coef_}x + {regressor.intercept_}")
```

```python
bias = regressor.score(x_train, y_train)
print(bias)
```

```python
variance = regressor.score(x_test, y_test)
print(variance)
```
